```python
import cv2
import matplotlib.pyplot as plt
import os

print("Danh sách file:", os.listdir('.'))

img = cv2.imread('image.jpg')  # hoặc 'Hinh.jpg' theo file tồn tại
if img is None:
    raise FileNotFoundError("Không tìm thấy ảnh image.jpg trong thư mục hiện tại")

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Dùng stop_data.xml (có sẵn) để phát hiện stop sign
stop_data = cv2.CascadeClassifier('stop_data.xml')
if stop_data.empty():
    raise FileNotFoundError("Không tìm thấy hoặc load không được stop_data.xml")

found = stop_data.detectMultiScale(img_gray, minSize=(20, 20))
print('Số đối tượng tìm được:', len(found))

if len(found) != 0:
    for (x, y, width, height) in found:
        cv2.rectangle(img_rgb, (x, y), (x + width, y + height), (0, 255, 0), 3)

plt.imshow(img_rgb)
plt.axis('off')
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import os

# Đọc ảnh
img = cv2.imread('hinh1.jpg')

if img is None:
    print("Không đọc được ảnh")
else:
    print("Đọc ảnh thành công")

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Kiểm tra file XML
print("XML tồn tại:", os.path.exists('stop_data.xml'))

stop_data = cv2.CascadeClassifier('stop_data.xml')

if stop_data.empty():
    print("Không load được stop_data.xml")
else:
    print("Load XML thành công")

# Detect
found = stop_data.detectMultiScale(img_gray, minSize=(20,20))

print("Số đối tượng phát hiện:", len(found))

# Vẽ khung
for (x, y, w, h) in found:
    cv2.rectangle(img_rgb, (x, y), (x+w, y+h), (0,255,0), 3)

# Hiển thị
plt.imshow(img_rgb)
plt.title("Stop Detection")
plt.axis('off')
plt.show()
```
