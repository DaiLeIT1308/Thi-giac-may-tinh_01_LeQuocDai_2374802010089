```python

# Import thư viện
import cv2 as cv
import matplotlib.pyplot as LeQuocDai
```

```python

# Đọc ảnh
img = cv.imread('Picture1.jpg')

# Kiểm tra kích thước ảnh
print("Kích thước ảnh:", img.shape)
```

```python

# Hiển thị ảnh (màu sẽ bị sai do OpenCV dùng BGR)
TenHoSV.imshow(img)
TenHoSV.title("Ảnh gốc (BGR)")
TenHoSV.axis('off')
TenHoSV.show()
```

### Nhận xét
Ảnh hiển thị sai màu vì OpenCV đọc ảnh theo thứ tự BGR thay vì RGB.

```python

# Cách 1: Chuyển BGR sang RGB
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

TenHoSV.imshow(img_rgb)
TenHoSV.title("Ảnh sau khi convert RGB")
TenHoSV.axis('off')
TenHoSV.show()
```

```python

# Phóng to ảnh
img_resize = cv.resize(img_rgb, None, fx=1.5, fy=1.5)

TenHoSV.imshow(img_resize)
TenHoSV.title("Ảnh phóng to")
TenHoSV.axis('off')
TenHoSV.show()
```

```python

# Crop ảnh (lấy vùng giữa)
h, w, _ = img_rgb.shape
crop = img_rgb[h//4:3*h//4, w//4:3*w//4]

TenHoSV.imshow(crop)
TenHoSV.title("Ảnh sau khi crop")
TenHoSV.axis('off')
TenHoSV.show()
```

```python

# Ghép ảnh ngang và dọc
img2 = cv.imread('Picture2.jpg')
img2 = cv.cvtColor(img2, cv.COLOR_BGR2RGB)
img2 = cv.resize(img2, (img_rgb.shape[1], img_rgb.shape[0]))

# Ghép ngang
concat_h = cv.hconcat([img_rgb, img2])
TenHoSV.imshow(concat_h)
TenHoSV.title("Ghép ngang")
TenHoSV.axis('off')
TenHoSV.show()

# Ghép dọc
concat_v = cv.vconcat([img_rgb, img2])
TenHoSV.imshow(concat_v)
TenHoSV.title("Ghép dọc")
TenHoSV.axis('off')
TenHoSV.show()
```
